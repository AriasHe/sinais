```python
def teste(t):
    if t >= -2 and t < -1:
        x_ = 1
    elif t >= -1 and t < 0:
        x_ = -2 * t
    elif t >= 0 and t < 1:
        x_ = 2 * t
    elif t>1 and t<2 :
        x_ = 1
    else:
        x_= 0
    return x_
```
