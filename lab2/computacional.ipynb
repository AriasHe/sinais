```python
import cmath as m
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
```

```python
def teste(t):
    if t >= -2 and t < -1:
        x_ = 1
    elif t >= -1 and t < 0:
        x_ = -2 * t
    elif t >= 0 and t < 1:
        x_ = 2 * t
    elif t>1 and t<2 :
        x_ = 1
    else:
        x_= 0
    return x_
```

```python
x1 = [-3,-2,-2,-1,-1,0,1,1,2,2,3]
y1 = [ 0,0, 1, 1, 2,0,2,1,1,0,0]
T=7
v = [-1,-10,-20,-50]
wo = 2 * m.pi*(1/7)
```

```python

for n in v:
    
    x_lista = []
    xlista = []
    ck_vetor = []
    w = []
    nneg = n*-1
    t_vetor = np.arange(-(T/2), (T/2), 0.001)

    
    for t in t_vetor:
        
        x_ = teste(t)
        x = 0 + 0j 
        ck_vetor.clear()
        w.clear()
        for k in range(n, nneg + 1):
            if k == 0:
                ck =4/7
                e = 1 
            else:
                ck = (-14 * (m.sin((k * m.pi) / 7) * m.sin((m.pi * k) / 7)) / (k**2*m.pi**2)) + ((m.sin((4 * m.pi * k) / 7) + m.sin((2 * m.pi * k) / 7)) / (k* m.pi))
                e = m.exp(1j * k * (2 * m.pi / 7) * t)
            ck_vetor.append(abs(ck))
            w.append(k)
            x = x + ck * e
        xlista.append(x)
        x_lista.append(x_)
    xarray = np.array(xlista)
    x_array = np.array(x_lista)
    grafico()

```

```python
def potencia(xarray,x_array,t_vetor):
    contagem = 0
    soma = 0
    for t in t_vetor:
        soma = (x_array[contagem]-xarray[contagem])**2 + soma
        contagem = contagem + 1
    soma = soma/len(xarray)
    soma = f"{soma.real:.2e}{soma.imag:+.2e}j"
    return (soma)
```
